### Matrix Algebra
* Matrix-vector mutiplication

```python
import numpy as np
```

```python
matrix =  np.array([[1,2,3],[4,5,6]])
vector = np.array([10,20,30])
matrix @ vector
```

* Matrix inverse

```python
A = np.array([[4,7],[2,6]])
A_inverse = np.linalg.inv(A)
print(A_inverse)
```
